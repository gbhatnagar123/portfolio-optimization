```python
import numpy as np

# Set random seed for reproducibility
rng = np.random.default_rng(0)

# Expected returns and asset volatilities
mu = np.array([0.05, 0.08, 0.12, 0.07])
vol = np.array([0.10, 0.15, 0.25, 0.12])

# Correlation matrix
Corr = np.array([
    [1, .3, .2, .1],
    [.3, 1, .4, .2],
    [.2, .4, 1, .3],
    [.1, .2, .3, 1]
])

# Create covariance matrix
Sigma = np.outer(vol, vol) * Corr

# Generate 2000 observations for 4 assets
R = rng.multivariate_normal(mu, Sigma, size=2000)

# Calculate sample covariance and average returns
S = np.cov(R.T)
m = R.mean(axis=0)
one = np.ones(4)

# Q3: Minimum variance portfolio
Si1 = np.linalg.solve(S, one)
w_min = Si1 / (one @ Si1)

print("Minimum Variance Weights:", w_min)
print("Portfolio Risk:", w_min @ S @ w_min)
print("Portfolio Return Variance:", (R @ w_min).var(ddof=1))

# Calculate constants for Q4 and Q5
a = one @ Si1
b = m @ Si1
c = m @ np.linalg.solve(S, m)

# Q4 and Q5: Calculate portfolios for different target returns
for r in np.linspace(0.04, 0.14, 6):

    lam = np.linalg.solve([[c, b], [b, a]], [r, 1])

    w = np.linalg.solve(S, lam[0] * m + lam[1] * one)

    print(
        "Target Return:", r,
        "Portfolio Variance:", w @ S @ w,
        "Parabola Variance:", (a*r**2 - 2*b*r + c) / (a*c - b**2)
    )
```

```python
# Q2: Check eigenvalues of the covariance matrix
eigenvalues = np.linalg.eigvalsh(S)

print("Covariance Matrix Eigenvalues:", eigenvalues)
print("All Eigenvalues Nonnegative:", np.all(eigenvalues >= -1e-10))
```

```python
# Repeat simulation with only 60 observations
R_60 = rng.multivariate_normal(mu, Sigma, size=60)


S_60 = np.cov(R_60.T)
m_60 = R_60.mean(axis=0)

Si1_60 = np.linalg.solve(S_60, one)
w_min_60 = Si1_60 / (one @ Si1_60)

risk_60 = w_min_60 @ S_60 @ w_min_60

print("60 Observations")
print("Estimated Returns:", m_60)
print("Minimum Variance Weights:", w_min_60)
print("Portfolio Variance:", risk_60)
```
