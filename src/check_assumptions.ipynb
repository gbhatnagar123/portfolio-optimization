```python

import numpy as np
import pandas as pd

from methods import my_estimate_mu_sigma, my_min_variance


# Load and clean real data
data = pd.read_csv(
    "../data/10_Industry_Portfolios.csv",
    skiprows=11,
    nrows=1203,
    index_col=0
)

data = data.apply(pd.to_numeric, errors="coerce")
data = data.replace([-99.99, -999], np.nan)
data = data.dropna()
data = data / 100

R = data.to_numpy()


# Estimate parameters and calculate weights
mu, Sigma = my_estimate_mu_sigma(R)
w = my_min_variance(Sigma)


print("REAL DATA ASSUMPTION CHECKS")

# 1. Missing and non-finite values
print("\nMissing Values:")
print(np.isnan(R).sum())

print("\nAll Returns Finite:")
print(np.isfinite(R).all())

# 2. Return ranges
print("\nMinimum Returns:")
print(R.min(axis=0))

print("\nMaximum Returns:")
print(R.max(axis=0))

# 3. Covariance matrix symmetry
print("\nCovariance Matrix Symmetric:")
print(np.allclose(Sigma, Sigma.T))

# 4. Positive definiteness
eigenvalues = np.linalg.eigvalsh(Sigma)

print("\nMinimum Eigenvalue:")
print(eigenvalues.min())

print("\nPositive Definite:")
print(eigenvalues.min() > 0)

# 5. Rank and condition number
print("\nCovariance Matrix Rank:")
print(np.linalg.matrix_rank(Sigma))

print("\nCondition Number:")
print(np.linalg.cond(Sigma))

# 6. Portfolio constraints
print("\nSum of Portfolio Weights:")
print(w.sum())

print("\nNegative Weights Present:")
print(np.any(w < 0))
```
