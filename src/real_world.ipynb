```python
import pandas as pd
import numpy as np
from methods import my_min_variance
```

```python
data = pd.read_csv(
    "../data/10_Industry_Portfolios.csv",
    skiprows=11,
    nrows=1203,
    index_col=0
)
```

```python
#Convert to decimal
data = data/100
```

```python
# Estimate monthly expected returns
mu_monthly = data.mean()

# Estimate monthly covariance matrix
Sigma_monthly = data.cov()

print("Monthly Expected Returns:")
print(mu_monthly)

print("\nMonthly Covariance Matrix:")
print(Sigma_monthly)
```

```python
mu_annual = mu_monthly * 12
Sigma_annual = Sigma_monthly * 12

print("Annualized Expected Returns:")
print(mu_annual)

print("\nAnnualized Covariance Matrix:")
print(Sigma_annual)
```

```python
w_real = my_min_variance(Sigma_annual)

print("Minimum-Variance Portfolio Weights:")
print(w_real)

print("\nSum of Weights:")
print(w_real.sum())
```

```python
print("Condition Number:")
print(np.linalg.cond(Sigma_annual))
```
